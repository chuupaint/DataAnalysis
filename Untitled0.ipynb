```python
import requests
import json


url = 'https://apis.data.go.kr/B552845/perDay/price'
api_key = '2998c904b551b2df2a537341342cacdf55c45a756123e80cbd1a9debe520a981'

params = {
    'serviceKey': api_key,
    'pageNo': '1',
    'numOfRows': '200',
    'returnType': 'json',
    'cond[exmn_ymd::GTE]': '20150101',
    'cond[exmn_ymd::LTE]': '20241231',
    'cond[se_cd::EQ]': '02',
    'cond[ctgry_cd::EQ]': '200',
    'cond[item_cd::EQ]': '211',
    'cond[grd_cd::EQ]': '04',
    'cond[sgg_cd::EQ]': '1101',
    'cond[mrkt_cd::EQ]': '0110211',
    'cond[vrty_cd::EQ]': '06'
}

all_items = []
page_no = 1
total_count = 0

while True:
    params['pageNo'] = str(page_no)

    response = requests.get(url, params=params)
    response.raise_for_status()

    data = response.json()

    # 사진에 나온 데이터 구조에 맞춰 추출
    body = data['response']['body']
    items = body['items']['item']
    total_count = int(body.get('totalCount', 0))

    # 데이터가 한 건일 경우에도 리스트로 처리
    if isinstance(items, dict):
        items = [items]

    # 현재 페이지 데이터 추가
    all_items.extend(items)

    print(f'{page_no}페이지 수집 완료: {len(all_items)}건')

    # 전체 데이터를 수집했으면 종료
    if len(all_items) >= total_count or len(items) == 0:
        break

    page_no += 1

# JSON 파일로 저장
file_name = f'배추_품종06.json'

with open(file_name, 'w', encoding='utf-8') as f:
    json.dump(all_items, f, ensure_ascii=False, indent=4)
```

```python
import json
import pandas as pd


with open('배추_통합.json', 'r', encoding='utf-8') as f:
    data = json.load(f)

df = pd.DataFrame(data)

df['exmn_ymd'] = pd.to_datetime(df['exmn_ymd'])

df['year'] = df['exmn_ymd'].dt.year
df['month'] = df['exmn_ymd'].dt.month

df['season'] = df['month'].map({
    3: 'spring', 4: 'spring', 5: 'spring',
    6: 'summer', 7: 'summer', 8: 'summer',
    9: 'fall', 10: 'fall', 11: 'fall',
    12: 'winter', 1: 'winter', 2: 'winter'
})

print(df[['exmn_ymd', 'year', 'month', 'season']].head(10))
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df['exmn_dd_prc'] = pd.to_numeric(df['exmn_dd_prc'])
year_price = df.groupby('year')['exmn_dd_prc'].mean().reset_index()
plt.figure(figsize=(12, 6))
sns.lineplot(x='year', y='exmn_dd_prc', data=year_price, marker='o')

plt.title('Average price change of cabbage - 3121')
plt.xlabel('Year', fontsize=12)
plt.ylabel('Price', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)
plt.savefig('연도별 배추 평균 가격 변화.png')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df['exmn_dd_prc'] = pd.to_numeric(df['exmn_dd_prc'])
season_price = df.groupby('season')['exmn_dd_prc'].mean().reset_index()

ax = sns.barplot(x='season', y='exmn_dd_prc', data=season_price)
ax.bar_label(ax.containers[0], fmt='%.1f')

plt.title('Seasonal Average Cabbage Price - 3121')
plt.xlabel('Year', fontsize=12)
plt.ylabel('Price', fontsize=12)
plt.show()
```
